```python
# 1. Mount your Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

import os
import zipfile
import subprocess

# 2. Setup paths
zip_path = "/content/drive/MyDrive/project.zip"
extract_path = "/content/project_work"

# 3. Clean and Unzip
if os.path.exists(zip_path):
    print("--- Unzipping project from Drive... ---")
    if not os.path.exists(extract_path):
        os.makedirs(extract_path)
        
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
    print("Extract complete.")
    
    # 4. SMART SEARCH: Find where 'make_video.py' actually is
    target_file = "make_video.py"
    found_dir = None
    
    for root, dirs, files in os.walk(extract_path):
        if target_file in files:
            found_dir = root
            break
    
    if found_dir:
        print(f"--- Found script in: {found_dir} ---")
        %cd {found_dir}
        
        # 5. Execute Rendering
        print("--- Starting Rendering Pipeline ---")
        !python make_video.py
    else:
        print(f"ERROR: Could not find '{target_file}' inside the zip.")
        print("Files found in extract folder:")
        !ls -R {extract_path}
else:
    print(f"ERROR: Could not find {zip_path}. Please check your Drive carefully!")
```
